```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("/content/demo_profile (2).csv")

fig, axes = plt.subplots(1, 2, figsize=(10, 4.2))
axes[0].plot(df.x_um, df.n_dark, label="dark", color="k")
axes[0].plot(df.x_um, df.n_illum, label="illuminated", color="tab:orange")
axes[0].set_title("Electron density: dark vs. illuminated")
axes[0].set_xlabel("x (µm)"); axes[0].set_ylabel("n (m$^{-3}$)")
axes[0].set_yscale("log"); axes[0].legend(); axes[0].grid(alpha=0.3)

axes[1].plot(df.x_um, df.p_dark, label="dark", color="k")
axes[1].plot(df.x_um, df.p_illum, label="illuminated", color="tab:orange")
axes[1].set_title("Hole density: dark vs. illuminated")
axes[1].set_xlabel("x (µm)"); axes[1].set_ylabel("p (m$^{-3}$)")
axes[1].set_yscale("log"); axes[1].legend(); axes[1].grid(alpha=0.3)

plt.tight_layout()
plt.show()
```
